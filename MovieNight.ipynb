##2. Data Structures: Lists & Dictionaries

```python
movies = [
    {"title": "The Matrix",        "year": 1999, "rating": 8.7, "genre": "Sci-Fi"},
    {"title": "Toy Story",         "year": 1995, "rating": 8.3, "genre": "Animation"},
    {"title": "Parasite",          "year": 2019, "rating": 8.5, "genre": "Thriller"},
    {"title": "Spirited Away",     "year": 2001, "rating": 8.6, "genre": "Animation"},
    {"title": "The Dark Knight",   "year": 2008, "rating": 9.0, "genre": "Action"},
    {"title": "Get Out",           "year": 2017, "rating": 7.7, "genre": "Thriller"},
    {"title": "La La Land",        "year": 2016, "rating": 8.0, "genre": "Musical"},
    {"title": "Interstellar",      "year": 2014, "rating": 8.7, "genre": "Sci-Fi"},
]

print("We have", len(movies), "movies loaded.")
movies[0]  # peek at the first one
```

```python
# 1) Print the title of the FIRST movie in the list.
print(movies[0]["title"])

# 2) Print the rating of the LAST movie in the list.
print(movies[-1]["rating"])
```

##3. Loops

```python
# Loop over every movie and print: "Title (year) — rating X"
for movie in movies:
  print(movie["title"], "(", movie["year"], ") - rating", movie["rating"])
```

##4. Conditionals

```python
# Loop through movies and print each title with its label.

for movie in movies:
  if movie["rating"] >= 9.0:
    print(movie["title"], "- Masterpiece")
  elif movie["rating"] >= 8.0:
    print(movie["title"], "- Great")
  else:
    print(movie["title"], "- Good")
```

##5. List Comprehension

```python
# 1) A list of ALL genres (may include duplicates).
all_genres = [movie["genre"] for movie in movies]
print("Genres:", all_genres)

# 2) A list of titles for movies rated 8.5 or higher.
top_titles = [movie["title"] for movie in movies if movie["rating"] >= 8.5]
print("Top rated:", top_titles)

# 3) A list of titles from movies made in the year 2000 or LATER.
this_century = [movie["title"] for movie in movies if movie ["year"] >= 2000]
print("2000 and later:", this_century)
```

##6. Functions

```python
def describe_movie(movie):
    # Should return a string like:
    # "The Matrix is a Sci-Fi film from 1999 rated 8.7"
    # Use the movie's "title", "genre", "year", and "rating" keys.
    return movie["title"] + " is a " + movie["genre"] + " film from " + str(movie["year"]) + " rated " + str(movie["rating"])
    pass


# Test it (this should print a full sentence, not "None"):
print(describe_movie(movies[0]))
print(describe_movie(movies[2]))
```

```python
def average_rating(movie_list):
    # Add up every movie's rating, then divide by how many there are.
    # Hint: len(movie_list) tells you how many movies there are.
    total = 0
    for movie in movie_list:
      total = total + movie["rating"]
    return total / len(movie_list)
    pass


print("Average rating:", average_rating(movies))
```

##7. Lambda Functions

```python
# Sort movies by rating, HIGHEST first (descending).
# Fill in the lambda so it returns each movie's rating.
movies_by_rating = sorted(movies, key=lambda m:m["rating"], reverse=True)  # replace ___

for m in movies_by_rating:
    print(m["rating"], m["title"])
```
